# K-Pick 재구매 예측 모델 학습
**데이터**: `data/prep/k-pick_total_v3.csv`  
**타깃**: `label` (1 = 다음 주문에 재구매, 0 = 미구매)  
**모델**: XGBoost / Logistic Regression / LightGBM

```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score, f1_score, roc_auc_score,
    classification_report, confusion_matrix
)

import xgboost as xgb
import lightgbm as lgb

import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

font_path = 'C:/Windows/Fonts/malgun.ttf'
font = fm.FontProperties(fname=font_path)
plt.rcParams['font.family'] = font.get_name()
plt.rcParams['axes.unicode_minus'] = False

RANDOM_STATE = 42
print('라이브러리 로드 완료')
```

## 1. 데이터 로드

```python
df = pd.read_csv('../data/prep/k-pick_total_v3.csv')
print('데이터 크기:', df.shape)
print('\nLabel 분포:')
print(df['label'].value_counts())
print(f'\n양성 비율: {df["label"].mean():.4f}')
df.head(3)
```

## 2. 피처 / 타깃 분리

```python
DROP_COLS = ['user_id', 'product_id', 'order_id', 'label']
FEATURE_COLS = [c for c in df.columns if c not in DROP_COLS]

X = df[FEATURE_COLS].copy()
y = df['label'].copy()

# NaN 처리 (중앙값 대체)
X = X.fillna(X.median())

print('피처 수:', len(FEATURE_COLS))
print('피처 목록:', FEATURE_COLS)
```

## 3. Train / Validation 분할

```python
X_train, X_val, y_train, y_val = train_test_split(
    X, y,
    test_size=0.2,
    random_state=RANDOM_STATE,
    stratify=y
)

print(f'학습 세트: {X_train.shape}  |  Label 분포: {y_train.mean():.4f}')
print(f'검증 세트: {X_val.shape}  |  Label 분포: {y_val.mean():.4f}')
```

---
## 4. Logistic Regression

```python
# 데이터가 크므로 샘플링 후 학습 (LR은 스케일링 필수)
SAMPLE_SIZE = 500_000
sample_idx = X_train.sample(n=min(SAMPLE_SIZE, len(X_train)), random_state=RANDOM_STATE).index
X_train_lr = X_train.loc[sample_idx]
y_train_lr = y_train.loc[sample_idx]

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_lr)
X_val_scaled   = scaler.transform(X_val)

print(f'LR 학습 샘플 수: {len(X_train_lr):,}')

lr_model = LogisticRegression(
    C=1.0,
    max_iter=300,
    class_weight='balanced',
    solver='saga',
    random_state=RANDOM_STATE,
    n_jobs=-1
)
lr_model.fit(X_train_scaled, y_train_lr)

lr_pred      = lr_model.predict(X_val_scaled)
lr_pred_prob = lr_model.predict_proba(X_val_scaled)[:, 1]

lr_acc = accuracy_score(y_val, lr_pred)
lr_f1  = f1_score(y_val, lr_pred)
lr_auc = roc_auc_score(y_val, lr_pred_prob)

print(f'[Logistic Regression]  Accuracy: {lr_acc:.4f}  |  F1: {lr_f1:.4f}  |  AUC: {lr_auc:.4f}')
print(classification_report(y_val, lr_pred, target_names=['미구매(0)', '재구매(1)']))
```

---
## 5. XGBoost

```python
scale_pos_weight = (y_train == 0).sum() / (y_train == 1).sum()
print(f'scale_pos_weight: {scale_pos_weight:.2f}')

xgb_model = xgb.XGBClassifier(
    n_estimators=1000,
    learning_rate=0.05,
    max_depth=6,
    min_child_weight=30,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_alpha=0.1,
    reg_lambda=0.1,
    scale_pos_weight=scale_pos_weight,
    tree_method='hist',
    eval_metric='auc',
    early_stopping_rounds=50,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

xgb_model.fit(
    X_train, y_train,
    eval_set=[(X_val, y_val)],
    verbose=100
)

xgb_pred      = xgb_model.predict(X_val)
xgb_pred_prob = xgb_model.predict_proba(X_val)[:, 1]

xgb_acc = accuracy_score(y_val, xgb_pred)
xgb_f1  = f1_score(y_val, xgb_pred)
xgb_auc = roc_auc_score(y_val, xgb_pred_prob)

print(f'\n[XGBoost]  Accuracy: {xgb_acc:.4f}  |  F1: {xgb_f1:.4f}  |  AUC: {xgb_auc:.4f}')
print(classification_report(y_val, xgb_pred, target_names=['미구매(0)', '재구매(1)']))
```

---
## 6. LightGBM

```python
lgb_train = lgb.Dataset(X_train, label=y_train)
lgb_val   = lgb.Dataset(X_val,   label=y_val, reference=lgb_train)

params = {
    'objective':         'binary',
    'metric':            'auc',
    'learning_rate':     0.05,
    'num_leaves':        63,
    'max_depth':         7,
    'min_child_samples': 30,
    'feature_fraction':  0.8,
    'bagging_fraction':  0.8,
    'bagging_freq':      5,
    'lambda_l1':         0.1,
    'lambda_l2':         0.1,
    'is_unbalance':      True,
    'verbose':           -1,
    'seed':              RANDOM_STATE,
}

lgb_model = lgb.train(
    params,
    lgb_train,
    num_boost_round=1000,
    valid_sets=[lgb_val],
    callbacks=[
        lgb.early_stopping(stopping_rounds=50),
        lgb.log_evaluation(period=100),
    ]
)

lgb_pred_prob = lgb_model.predict(X_val)
lgb_pred      = (lgb_pred_prob >= 0.5).astype(int)

lgb_acc = accuracy_score(y_val, lgb_pred)
lgb_f1  = f1_score(y_val, lgb_pred)
lgb_auc = roc_auc_score(y_val, lgb_pred_prob)

print(f'\n[LightGBM]  Accuracy: {lgb_acc:.4f}  |  F1: {lgb_f1:.4f}  |  AUC: {lgb_auc:.4f}')
print(classification_report(y_val, lgb_pred, target_names=['미구매(0)', '재구매(1)']))
```

---
## 7. 모델 성능 비교

```python
results = pd.DataFrame({
    '모델':      ['Logistic Regression', 'XGBoost', 'LightGBM'],
    'Accuracy': [lr_acc, xgb_acc, lgb_acc],
    'F1 Score': [lr_f1,  xgb_f1,  lgb_f1],
    'AUC':      [lr_auc, xgb_auc, lgb_auc],
}).set_index('모델')

print('=== 모델 성능 비교 ===')
print(results.round(4))

fig, axes = plt.subplots(1, 3, figsize=(13, 4))
metrics = ['Accuracy', 'F1 Score', 'AUC']
colors  = ['steelblue', 'tomato', 'seagreen']

for ax, metric, color in zip(axes, metrics, colors):
    ax.bar(results.index, results[metric], color=color, width=0.5)
    ax.set_title(metric, fontsize=13)
    ax.set_ylim(0, 1)
    ax.set_xticklabels(results.index, rotation=15, ha='right')
    for i, v in enumerate(results[metric]):
        ax.text(i, v + 0.01, f'{v:.4f}', ha='center', fontsize=9)

plt.suptitle('모델 성능 비교', fontsize=15, fontweight='bold')
plt.tight_layout()
plt.savefig('../data/prep/model_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
print('저장 완료 → ../data/prep/model_comparison.png')
```

## 8. 피처 중요도 (XGBoost / LightGBM)

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# XGBoost
xgb_imp = pd.Series(
    xgb_model.feature_importances_, index=FEATURE_COLS
).sort_values(ascending=True)
xgb_imp.plot(kind='barh', ax=axes[0], color='tomato')
axes[0].set_title('XGBoost 피처 중요도', fontsize=13)
axes[0].set_xlabel('importance')

# LightGBM
lgb_imp = pd.Series(
    lgb_model.feature_importance(importance_type='gain'),
    index=FEATURE_COLS
).sort_values(ascending=True)
lgb_imp.plot(kind='barh', ax=axes[1], color='steelblue')
axes[1].set_title('LightGBM 피처 중요도 (Gain)', fontsize=13)
axes[1].set_xlabel('importance')

plt.tight_layout()
plt.savefig('../data/prep/feature_importance.png', dpi=150, bbox_inches='tight')
plt.show()
print('저장 완료 → ../data/prep/feature_importance.png')
```
